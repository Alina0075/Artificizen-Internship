### Question 1

```python
class Animal:
    def speak(self):
        print("Animal sound")

class Dog(Animal):
    def speak(self):
        print("Dog Bark")

class Cat(Animal):
    def speak(self):
        print("Cat Meow")

Dog().speak()
Cat().speak()
```

### Question 2

```python
class InsufficientBalanceError(Exception):
    pass
class BankAccount:
    def __init__(self, balance, deposit, withdrawl):
        self.balance = balance
        self.deposit = deposit
        self.withdrawl = withdrawl

    def dep(self):
        self.balance = self.balance + self.deposit
        print(f'Updated Balance is {self.balance}')

    def withdr(self):
        if self.balance > self.withdrawl:
            self.balance = self.balance - self.withdrawl
            print(f'Updated Balance is {self.balance}')
        else:
            raise InsufficientBalanceError("Insufficient Balance")
obj2 = BankAccount(10000, 3000, 2000)
obj3 = BankAccount(5000, 0, 6000)

obj2.dep()

try:
    obj3.withdr()
except InsufficientBalanceError as e:
    print(e)
```

### Question 5

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __str__(self):
        return f"Point({self.x}, {self.y})"

    def __repr__(self):
        return f"Point(x={self.x}, y={self.y})"
p = Point(3, 5)
print(p)
print(repr(p))
```
