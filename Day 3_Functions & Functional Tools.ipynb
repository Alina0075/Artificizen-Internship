### Question 4

```python
def factorial(num):
    if num<=1:
        return 1
    else:
        return (num* factorial(num-1))
fac=factorial(5)
print(f'Factorial of 5 is {fac}')
```

### Question 5

```python
students = [
    {"name": "Alina",  "marks": 92},
    {"name": "Jazib",  "marks": 78},
    {"name": "Haseeb", "marks": 85},
    {"name": "Zara",   "marks": 95},
]
asc=sorted(students, key = lambda a: a['marks'])
desc=sorted(students, key = lambda d: d['marks'], reverse=True)

print(f'Sorted in Ascending order {asc}')
print(f'Sorted in Descending Order {desc}')
```

### Question 6

```python
# Default argument values are evaluated once when the function is defined, not each time it is called. A mutable default (list, dict, set) 
#is therefore shared across ALL calls and hence mutations persist!

def add_item_bad(item, basket=[]):   # dangerous!
    basket.append(item)
    return basket

print(add_item_bad("apple"))   # ['apple'] correct
print(add_item_bad("mango"))   # ['apple', 'mango'] unexpected!
# The same [] object is reused on every call.

def add_item_good(item, basket=None):   # correct
    # None is immutable, so no shared state problem.
    # Create a fresh list on each call when no basket is provided.
    if basket is None:
        basket = []
    basket.append(item)
    return basket

print(add_item_good("apple"))   # ['apple']
print(add_item_good("mango"))   # ['mango']  fresh list
```
