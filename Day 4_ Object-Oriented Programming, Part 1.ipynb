### Question 2

```python
class BankAccount:
    def __init__(self, balance, deposit, withdrawl):
        self.balance = balance
        self.deposit = deposit
        self.withdrawl = withdrawl

    def dep(self):
        self.balance= self.balance+ self.deposit
        print(f'Updated Balance is {self.balance}')
    def withdr(self):
        if self.balance > self.withdrawl:
            self.balance= self.balance- self.withdrawl
            print(f'Updated Balance is {self.balance}')
        else:
            print('Insufficent Balance')
            
obj2=BankAccount(10000,3000,2000)
obj3=BankAccount(5000,0,6000)
obj2.dep()
obj3.withdr()
```

```python
class Animal:
    count=0
    def __init__(self,animal1):
        self.animal1=animal1
        Animal.count=Animal.count+1
        
    def display(self):
        print(f'My Faviourte animal is: {self.animal1}')

    @classmethod
    def total_objects(cls):
        print(f"Total objects created: {cls.count}")
        
obj1=Animal('Cat')
obj2=Animal('bunny')
obj1.display()
obj2.display()
Animal.total_objects()
```

### Question 4

```python
class BankBalance:
    def __init__(self, balance):
        self.balance = balance      

    @property
    def balance(self):
        return self.__balance
    @balance.setter
    def balance(self, value):
        if value < 0:
            print("Balance can't be negative")
        else:
            self.__balance = value


obj1 = BankBalance(1000)
print("Balance:", obj1.balance)

obj1.balance = 2000
print("Updated Balance:", obj1.balance)
```

### Question 5

```python
class Person:
    def __init__(self,name,age):
        self.name=name
        self.age=age

    @classmethod
    def birth_year(cls,name,birth_year):
        current_year = 2026
        age = current_year - birth_year
        return cls(name, age)

    def display(self):
        print(f"Name: {self.name}")
        print(f"Age: {self.age}")

obj1=Person('Alina',30)
obj2=Person.birth_year('Fiza',1999)

obj1.display()
obj2.display()
```
