### Question 1

```python
num = int(input("Enter a number: "))
isprime = True

if num<=1:
    print('Number is not prime: ')
else:
    for i in range(2,int(num**0.5)+1): #can use simply num but would increase the time complexity
        if num%i==0:
            isprime = False
            break
    if (isprime==True):
        print('Number is prime')
    elif isprime==False:
        print('Number is not prime: ')
```

```python
numbers=[1,2,3,0,10,20,32,15]
minn=numbers[0]
maxx=numbers[0]

for i in range(len(numbers)):
    if maxx<numbers[i]:
        maxx=numbers[i]
    elif minn>numbers[i]:
        minn=numbers[i]

print(f'Maximum Number is: {maxx}')
print(f'Minimum Number is: {minn}')
```
