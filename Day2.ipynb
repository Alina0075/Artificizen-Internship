```python
def most_frequent(lst):
    dict1={}
    for i in lst:
        if i in dict1:
            dict1[i]=dict1[i]+1
        else:
            dict1[i]=1
    maxx=0
    maxx_val=0
    for key,value in dict1.items():
        if maxx<value:
            maxx=value
            maxx_val=key
    return maxx_val
#if we use max function just need to add this in return line    return max(dict1, key=dict1.get)
numbers = [1,2,3,2,4,2,5,3,3,3]
print('Most frequent number in the list is: ')
print(most_frequent(numbers))
```
